```python
def voronoi_finite_polygons(vor, bounding_box):
    """
    scipy Voronoi output-কে finite (bounded) shapely polygon এ convert করে
    bounding_box বাইরে চলে যাওয়া ridge গুলো bounding box দিয়ে clip করা হয়
    """
    new_regions = []
    center = vor.points.mean(axis=0)
    radius = max(bounding_box[2] - bounding_box[0], bounding_box[3] - bounding_box[1]) * 2

    # প্রতিটা point এর জন্য কোন কোন ridge সংযুক্ত সেটা ম্যাপ করা
    all_ridges = {}
    for (p1, p2), (v1, v2) in zip(vor.ridge_points, vor.ridge_vertices):
        all_ridges.setdefault(p1, []).append((p2, v1, v2))
        all_ridges.setdefault(p2, []).append((p1, v1, v2))

    for p1, region_index in enumerate(vor.point_region):
        vertices = vor.regions[region_index]

        if all(v >= 0 for v in vertices):
            new_regions.append([vor.vertices[v] for v in vertices])
            continue

        ridges = all_ridges[p1]
        new_region_vertices = [vor.vertices[v] for v in vertices if v >= 0]

        for p2, v1, v2 in ridges:
            if v2 < 0:
                v1, v2 = v2, v1
            if v1 >= 0:
                continue

            t = vor.points[p2] - vor.points[p1]
            t = t / np.linalg.norm(t)
            n = np.array([-t[1], t[0]])

            midpoint = vor.points[[p1, p2]].mean(axis=0)
            direction = np.sign(np.dot(midpoint - center, n)) * n
            far_point = vor.vertices[v2] + direction * radius

            new_region_vertices.append(far_point)

        # Sort vertices by angle (so polygon isn't self-intersecting)
        vertices_arr = np.array(new_region_vertices)
        angles = np.arctan2(vertices_arr[:,1] - center[1], vertices_arr[:,0] - center[0])
        new_region_vertices = vertices_arr[np.argsort(angles)]

        new_regions.append(new_region_vertices)

    return new_regions
```

```python
def voronoi_areas_with_names(gdf, region_polys, region_pts, crs_from='EPSG:4326', crs_to='EPSG:32646'):
    rows = []
    for region_id, poly in region_polys.items():
        pt_indices = region_pts[region_id]
        for pt_idx in pt_indices:
            name = gdf.iloc[pt_idx]['name']
            rows.append({'name': name, 'geometry': poly})
    
    result = gpd.GeoDataFrame(rows, geometry='geometry', crs=crs_from)
    result_proj = result.to_crs(crs_to)
    result['area_km2'] = result_proj.geometry.area / 1_000_000
    return result
```

```python
import numpy as np
from scipy.spatial import cKDTree

# 1. Grid points toiri kora (Dhaka bbox er upor, ~0.005 degree spacing, mota mota ~500m grid)
grid_lon = np.arange(lon_min, lon_max, 0.005)
grid_lat = np.arange(lat_min, lat_max, 0.005)
grid_points = np.array([[lo, la] for lo in grid_lon for la in grid_lat])

# 2. Protyek category-r jonno KDTree banano (fast nearest-neighbor search-er jonno)
def nearest_distance_km(gdf, grid_points):
    coords = np.array(list(zip(gdf.geometry.x, gdf.geometry.y)))
    tree = cKDTree(coords)
    dist_deg, _ = tree.query(grid_points)
    # degree theke approximate km convert (Dhaka latitude-e 1 degree ~ 111 km)
    return dist_deg * 111

dist_hospital = nearest_distance_km(hospitals, grid_points)
dist_pharmacy = nearest_distance_km(pharmacies, grid_points)
dist_police = nearest_distance_km(police, grid_points)
dist_school = nearest_distance_km(schools, grid_points)

# 3. Combine kore ekta DataFrame banano
grid_df = pd.DataFrame({
    'lon': grid_points[:, 0],
    'lat': grid_points[:, 1],
    'dist_hospital_km': dist_hospital,
    'dist_pharmacy_km': dist_pharmacy,
    'dist_police_km': dist_police,
    'dist_school_km': dist_school,
})

# 4. Composite "underserved score" - sob category-r distance-er average
grid_df['avg_distance_km'] = grid_df[['dist_hospital_km','dist_pharmacy_km','dist_police_km','dist_school_km']].mean(axis=1)

# 5. Shobcheye underserved 15ta zone dekho
print(grid_df.sort_values('avg_distance_km', ascending=False).head(15)[['lat','lon','dist_hospital_km','dist_pharmacy_km','dist_police_km','dist_school_km','avg_distance_km']].to_string(index=False))
```

```python
import numpy as np

grid_lon = np.arange(lon_min, lon_max, 0.005)
grid_lat = np.arange(lat_min, lat_max, 0.005)
grid_points = np.array([[lo, la] for lo in grid_lon for la in grid_lat])

print("Total grid points:", len(grid_points))
```

```python
from scipy.spatial import cKDTree

def nearest_distance_km(gdf, grid_points):
    coords = np.array(list(zip(gdf.geometry.x, gdf.geometry.y)))
    tree = cKDTree(coords)
    dist_deg, _ = tree.query(grid_points)
    return dist_deg * 111

dist_hospital = nearest_distance_km(hospitals, grid_points)
dist_pharmacy = nearest_distance_km(pharmacies, grid_points)
dist_police = nearest_distance_km(police, grid_points)
dist_school = nearest_distance_km(schools, grid_points)
```

```python
import pandas as pd

grid_df = pd.DataFrame({
    'lon': grid_points[:, 0],
    'lat': grid_points[:, 1],
    'dist_hospital_km': dist_hospital,
    'dist_pharmacy_km': dist_pharmacy,
    'dist_police_km': dist_police,
    'dist_school_km': dist_school,
})

grid_df['avg_distance_km'] = grid_df[['dist_hospital_km','dist_pharmacy_km','dist_police_km','dist_school_km']].mean(axis=1)

top_underserved = grid_df.sort_values('avg_distance_km', ascending=False).head(15)
print(top_underserved[['lat','lon','dist_hospital_km','dist_pharmacy_km','dist_police_km','dist_school_km','avg_distance_km']].to_string(index=False))
```

```python
# Threshold ভিত্তিতে underserved zone শ্রেণীবদ্ধ করা
def classify_zone(dist):
    if dist < 0.5:
        return "Well Served"
    elif dist < 1.0:
        return "Moderately Served"
    elif dist < 2.0:
        return "Underserved"
    else:
        return "Severely Underserved"

grid_df['zone_category'] = grid_df['avg_distance_km'].apply(classify_zone)

# Summary table
summary_table = grid_df['zone_category'].value_counts().reset_index()
summary_table.columns = ['Category', 'Grid Point Count']
summary_table['Percentage'] = (summary_table['Grid Point Count'] / len(grid_df) * 100).round(2)

print(summary_table.to_string(index=False))

# প্রতিটা individual category (hospital/pharmacy/police/school) এর জন্যও mean distance
print("\n=== Mean Distance by Service Type ===")
print(f"Hospital: {grid_df['dist_hospital_km'].mean():.3f} km")
print(f"Pharmacy: {grid_df['dist_pharmacy_km'].mean():.3f} km")
print(f"Police: {grid_df['dist_police_km'].mean():.3f} km")
print(f"School: {grid_df['dist_school_km'].mean():.3f} km")
```

```python
import matplotlib.pyplot as plt
from scipy.spatial import Voronoi, voronoi_plot_2d
import numpy as np

fig, axes = plt.subplots(2, 2, figsize=(18, 18))

categories = [
    ('dist_hospital_km', 'Hospital', hospitals, axes[0, 0]),
    ('dist_pharmacy_km', 'Pharmacy', pharmacies, axes[0, 1]),
    ('dist_police_km', 'Police', police, axes[1, 0]),
    ('dist_school_km', 'School', schools, axes[1, 1]),
]

vmax_common = grid_df[['dist_hospital_km','dist_pharmacy_km','dist_police_km','dist_school_km']].max().max()

for dist_col, name, facility_gdf, ax in categories:
    # Underserved heatmap (background layer)
    scatter = ax.scatter(grid_df['lon'], grid_df['lat'], c=grid_df[dist_col],
                          cmap='Reds', s=25, marker='s', vmin=0, vmax=vmax_common, alpha=0.7, zorder=1)

    # Voronoi diagram overlay (facility coverage boundaries)
    coords = np.array(list(zip(facility_gdf.geometry.x, facility_gdf.geometry.y)))
    if len(coords) >= 4:  # Voronoi needs at least 4 points
        vor = Voronoi(coords)
        voronoi_plot_2d(vor, ax=ax, show_points=False, show_vertices=False,
                         line_colors='blue', line_width=0.6, line_alpha=0.5, point_size=0)

    # Facility points (top layer)
    ax.scatter(facility_gdf.geometry.x, facility_gdf.geometry.y, c='black', s=8, marker='+', zorder=3, label=name)

    # Boundary ঠিক রাখা (Voronoi ইনফিনিট এজ চলে গেলে zoom out হয়ে যায়)
    ax.set_xlim(lon_min, lon_max)
    ax.set_ylim(lat_min, lat_max)

    plt.colorbar(scatter, ax=ax, label='Distance (km)')
    ax.set_title(f'{name}: Voronoi Coverage + Underserved Zones (n={len(facility_gdf)})')
    ax.set_xlabel('Longitude')
    ax.set_ylabel('Latitude')

plt.tight_layout()
plt.savefig('voronoi_combined_underserved.png', dpi=300)
plt.show()
```

```python
# প্রতিটা grid point একটা cell represent করে — cell size হিসাব করা
# grid spacing 0.005 degree ব্যবহার করেছিলে
grid_spacing_deg = 0.005

# Dhaka এর latitude (~23.8) এ degree-to-km conversion সামান্য ভিন্ন lat/lon এ
# 1 degree latitude ≈ 111 km (সব জায়গায় প্রায় same)
# 1 degree longitude ≈ 111 * cos(latitude) km
import numpy as np
avg_lat = grid_df['lat'].mean()
km_per_deg_lat = 111.0
km_per_deg_lon = 111.0 * np.cos(np.radians(avg_lat))

cell_width_km = grid_spacing_deg * km_per_deg_lon
cell_height_km = grid_spacing_deg * km_per_deg_lat
cell_area_km2 = cell_width_km * cell_height_km

print(f"প্রতিটা grid cell এর আয়তন: {cell_area_km2:.4f} km²")

# প্রতিটা zone category এর মোট area হিসাব করা
area_summary = grid_df['zone_category'].value_counts().reset_index()
area_summary.columns = ['Category', 'Grid Point Count']
area_summary['Approx Area (km²)'] = (area_summary['Grid Point Count'] * cell_area_km2).round(2)
area_summary['Percentage'] = (area_summary['Grid Point Count'] / len(grid_df) * 100).round(2)

print("\n=== Underserved Area Summary (with km²) ===")
print(area_summary.to_string(index=False))

total_area_km2 = len(grid_df) * cell_area_km2
print(f"\nমোট বিশ্লেষণকৃত এলাকা: {total_area_km2:.2f} km²")
```

```python
from geopy.geocoders import Nominatim
import time

geolocator = Nominatim(user_agent="dhaka_thesis_priority_zones")

def get_area_name(lat, lon):
    try:
        location = geolocator.reverse((lat, lon), language='en')
        if location:
            # শুধু প্রথম ২-৩টা address component নেওয়া (পুরো ঠিকানা না, সংক্ষিপ্ত)
            parts = location.address.split(',')
            return ', '.join(parts[:2]).strip()
        return "Unknown"
    except Exception as e:
        return f"Error"

# Top 10 severely underserved zone নেওয়া
priority_zones = grid_df.sort_values('avg_distance_km', ascending=False).head(10).copy()

priority_zones['area_name'] = priority_zones.apply(
    lambda row: get_area_name(row['lat'], row['lon']), axis=1
)
time.sleep(0)  # rate limit এর জন্য নিচে loop এ delay দেওয়া আছে

# কোন সার্ভিস সবচেয়ে দূরে (most critical shortage) সেটাও বের করা
def most_lacking_service(row):
    dists = {
        'Hospital': row['dist_hospital_km'],
        'Pharmacy': row['dist_pharmacy_km'],
        'Police': row['dist_police_km'],
        'School': row['dist_school_km'],
    }
    return max(dists, key=dists.get)

priority_zones['most_critical_gap'] = priority_zones.apply(most_lacking_service, axis=1)

# ফাইনাল টেবিল
priority_table = priority_zones[['area_name', 'lat', 'lon', 'avg_distance_km', 'most_critical_gap']].copy()
priority_table.columns = ['Area', 'Latitude', 'Longitude', 'Avg Distance (km)', 'Most Critical Gap']
priority_table['Avg Distance (km)'] = priority_table['Avg Distance (km)'].round(3)

print("=== Top 10 Priority Zones for New Facility Placement ===")
print(priority_table.to_string(index=False))
```
